# Credit Risk Prediction - Exploratory Data Analysis (EDA)

## 1. Problem Statement
In lending, predicting whether an applicant is likely to default (`loan_status = 1`) or repay their loan (`loan_status = 0`) is essential to mitigate credit loss.
This notebook analyzes the `Credit Risk Dataset` to understand distributions, relationships, missing values, and outliers before model training.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset from local data folder
df = pd.read_csv('../data/credit_risk_dataset.csv')
print(f"Dataset Shape: {df.shape[0]} rows, {df.shape[1]} columns")
df.head()
```

## 2. Data Types and Missing Values

```python
print("--- Missing Values ---")
print(df.isnull().sum())
print(f"\nDuplicate Rows: {df.duplicated().sum()}")
```

## 3. Target Distribution (`loan_status`)
- **0 = Non-Default (Low Risk)**
- **1 = Default (High Risk)**

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='loan_status', palette='coolwarm')
plt.title("Target Distribution: Non-Default vs Default")
plt.show()
print(df['loan_status'].value_counts(normalize=True))
```

## 4. Detecting Outliers in Age and Employment Length
Notice extreme domain errors (e.g. `person_age` > 100 or `person_emp_length` > 100).

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.boxplot(data=df, x='person_age', ax=axes[0], color='skyblue')
axes[0].set_title("Distribution of Person Age")

sns.boxplot(data=df, x='person_emp_length', ax=axes[1], color='salmon')
axes[1].set_title("Distribution of Employment Length")
plt.show()
```

## 5. Loan Intent & Home Ownership vs Default Rate

```python
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='loan_intent', hue='loan_status', palette='Set2')
plt.title("Loan Intent vs Loan Status")
plt.xticks(rotation=30)
plt.show()
```

## 6. Correlation Matrix (Numerical Features)

```python
plt.figure(figsize=(10, 8))
numeric_df = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title("Correlation Heatmap")
plt.show()
```
